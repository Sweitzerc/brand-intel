## Step 1 · Your settings
Change the boxes on the right if you like, then keep going. To change them later, edit the boxes and run **Runtime → Run all** again.

```python
#@title Settings { display-mode: "form" }
brand = "Canes Galore"  #@param {type:"string"}
competitors = "Fashionable Canes, HurryCane"  #@param {type:"string"}
my_domain = "canesgalore.com"  #@param {type:"string"}
```

## Step 3 · Start the dashboard
The dashboard appears below this step. To stop it, click the ⏹ button next to this step.

```python
#@title Start the dashboard { display-mode: "form" }
import sys, yaml

# Save the settings from Step 1 into the project's config file.
path = "/content/brand-intel/config.yaml"
cfg = yaml.safe_load(open(path))
cfg["brand"] = brand.strip()
cfg["competitors"] = [c.strip() for c in competitors.split(",") if c.strip()]
cfg["domains"] = [my_domain.strip()] if my_domain.strip() else []
yaml.safe_dump(cfg, open(path, "w"), sort_keys=False)

# Forget any copy loaded by an earlier run so new settings take effect.
for name in [m for m in sys.modules if m == "app" or m.startswith("brand_intel")]:
    del sys.modules[name]
sys.path.insert(0, "/content/brand-intel")

import gradio as gr
gr.close_all()
import app
app.demo.launch(share=True, debug=True)
```
